# Logistic regression

Logistic regression predicts a **category** (here, class 0 vs. class 1) by
fitting an S-shaped probability curve. It's the classification counterpart to
the [linear regression](linear-regression.ipynb) you just saw, using
[`linfa-logistic`](https://docs.rs/linfa-logistic).

```rust
:dep ndarray = { version = "0.15" }
:dep linfa = { version = "0.7" }
:dep linfa-logistic = { version = "0.7" }
use ndarray::{array, Array1};

// One feature; classes are cleanly separated: low x -> class 0, high x -> 1.
let x = array![[1.0_f64], [2.0], [3.0], [6.0], [7.0], [8.0]];
let y: Array1<usize> = array![0, 0, 0, 1, 1, 1];
println!("{} samples in classes {{0, 1}}", x.nrows());
```

## Fit and predict

As with linear regression, we fit inside a `{ }` block. `predict` returns an
`Array1<usize>` of class labels, which *is* a nameable type, so we can let it
escape the block and reuse it below.

```rust
use linfa::prelude::*;
use linfa::Dataset;
use linfa_logistic::LogisticRegression;
use ndarray::Array1;

// Explicit `: Array1<usize>` so evcxr can persist predictions across cells.
let predictions: Array1<usize> = {
    let dataset = Dataset::new(x.clone(), y.clone());
    let model = LogisticRegression::default()
        .max_iterations(200)
        .fit(&dataset)
        .expect("fit failed");
    model.predict(&dataset)
};
println!("predicted: {:?}", predictions);
println!("actual   : {:?}", y);
```

## Accuracy

Compare predictions to the true labels:

```rust
let correct = predictions
    .iter()
    .zip(y.iter())
    .filter(|(p, t)| p == t)
    .count();
let accuracy = correct as f64 / y.len() as f64;
println!("accuracy = {:.0}% ({}/{})", accuracy * 100.0, correct, y.len());
```
