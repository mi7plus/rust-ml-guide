# Setup: running the environment

Every chapter in this guide is a live Rust notebook running on the
[`evcxr`](https://github.com/evcxr/evcxr) Jupyter kernel, inside a Docker image
that ships a pinned Rust toolchain and a pre-warmed cache of the ML crates.

## Quickstart

From the repository root:

```bash
make build   # build the Docker image (first time only, ~15-20 min)
make up      # start JupyterLab at http://localhost:8888
```

Open any chapter notebook and pick the **Rust** kernel. That's it — no local
Rust or Python install required.

## Confirm the kernel works

The cell below is real Rust, compiled and run by the kernel:

````rust
let greeting = "Rust kernel is alive";
println!("{}", greeting);
// The value of the last expression is displayed as the cell output:
2 + 2
````

```{note}
The first cell that pulls a crate (`:dep ...`) compiles it. Thanks to the
pre-warmed cache this is seconds, not minutes. See the
[crate reference](../appendix/crate-reference.md) for the full stack and
troubleshooting tips.
```
