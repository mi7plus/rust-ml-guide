# Decision trees

A decision tree classifies by asking a sequence of yes/no questions about the
features — splitting the data to make each group as pure as possible. The result
is a model you can read as if/else rules. We use
[`linfa-trees`](https://docs.rs/linfa-trees).

This is supervised classification, like [logistic
regression](../02-regression/logistic-regression.ipynb), so we need labelled
data: features `x` and integer class labels `y`.

```rust
:dep ndarray = { version = "0.15" }
:dep linfa = { version = "0.7" }
:dep linfa-trees = { version = "0.7" }
use ndarray::{array, Array1};

// Two features, two classes.
let x = array![
    [1.0_f64, 1.0], [1.2, 0.9], [0.9, 1.1],
    [5.0, 5.0], [5.1, 4.9], [4.8, 5.2]
];
let y: Array1<usize> = array![0, 0, 0, 1, 1, 1];
println!("{} samples, {} features", x.nrows(), x.ncols());
```

```rust
use linfa::prelude::*;
use linfa::Dataset;
use linfa_trees::DecisionTree;
use ndarray::Array1;

// Explicit `: Array1<usize>` so evcxr can persist predictions across cells.
let predictions: Array1<usize> = {
    let dataset = Dataset::new(x.clone(), y.clone());
    let model = DecisionTree::params()
        .fit(&dataset)
        .expect("tree fit failed");
    model.predict(&dataset)
};
println!("predicted: {:?}", predictions);
println!("actual   : {:?}", y);
```

## Accuracy

On this cleanly separable data a single split is enough for a perfect fit:

```rust
let correct = predictions
    .iter()
    .zip(y.iter())
    .filter(|(p, t)| p == t)
    .count();
println!("accuracy = {:.0}% ({}/{})", correct as f64 / y.len() as f64 * 100.0, correct, y.len());
```
