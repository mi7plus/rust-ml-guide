# ndarray basics

[`ndarray`](https://docs.rs/ndarray) is Rust's N-dimensional array crate — the
equivalent of NumPy. Every ML crate in this guide speaks `ndarray`, so it's the
foundation for everything that follows.

```{note}
We pin `ndarray = "0.15"` throughout the book because `linfa` 0.7 depends on
that version. See the [crate reference](../appendix/crate-reference.md) for why
mixing versions breaks things.
```

````rust
:dep ndarray = { version = "0.15" }
use ndarray::array;

// A 1-D array (vector) and a 2-D array (matrix):
let v = array![1.0_f64, 2.0, 3.0];
let m = array![[1.0_f64, 2.0], [3.0, 4.0]];
println!("vector v = {}", v);
println!("matrix m =\n{}", m);
println!("m shape = {:?}", m.dim());
````

## Element-wise arithmetic

Operators broadcast element-wise, and scalars broadcast across the whole array —
just like NumPy:

````rust
use ndarray::array;
let a = array![[1.0_f64, 2.0], [3.0, 4.0]];
let b = array![[10.0_f64, 20.0], [30.0, 40.0]];
println!("a + b =\n{}", &a + &b);
println!("a * 2 =\n{}", &a * 2.0);
println!("a .* b (element-wise) =\n{}", &a * &b);
````

## Linear algebra & reductions

`.dot()` is matrix multiplication, `.t()` is transpose, and reductions like
`.sum()` and `.mean()` collapse the array to a scalar:

````rust
use ndarray::array;
let a = array![[1.0_f64, 2.0], [3.0, 4.0]];
let b = array![[5.0_f64, 6.0], [7.0, 8.0]];
println!("a . b (matmul) =\n{}", a.dot(&b));
println!("a transposed =\n{}", a.t());
println!("sum = {}", a.sum());
println!("mean = {}", a.mean().unwrap());
println!("first row = {}", a.row(0));
println!("second column = {}", a.column(1));
````
